```python
import pandas as pd
import urllib.parse
from sqlalchemy import create_engine

# 1. Database Configuration
raw_password = "Your Password Here"
encoded_password = urllib.parse.quote_plus(raw_password)
connection_str = f'mysql+mysqlconnector://root:{encoded_password}@localhost/olist_ecommerce'
engine = create_engine(connection_str)

# 2. File Paths Definition
# Ensure these paths are accurate and accessible on your system
file_paths = {
    'order_payments': r"C:\Users\dell\Downloads\archive (2)\olist_order_payments_dataset.csv",
    'order_reviews': r"C:\Users\dell\Downloads\archive (2)\olist_order_reviews_dataset.csv",
    'orders': r"C:\Users\dell\Downloads\archive (2)\olist_orders_dataset - Copy.csv"
}

# 3. Batch Ingestion Process
print("Starting batch ingestion process...")

for table_name, path in file_paths.items():
    print(f"Processing: {table_name}...")
    
    # Load dataset
    df = pd.read_csv(path)
    
    # Ingest into MySQL
    df.to_sql(table_name, con=engine, if_exists='replace', index=False, chunksize=20000)
    print(f"    [✓] Table '{table_name}' successfully imported.")

print("\nProcess completed: All datasets have been successfully ingested into the database.")
```
